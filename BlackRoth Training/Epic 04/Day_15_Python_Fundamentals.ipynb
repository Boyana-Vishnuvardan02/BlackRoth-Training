```python
#Create a function that accepts sales values and returns the total sales and average sales.
```

```python
def sales_analysis(sales):
    total = sum(sales)
    average = total/len(sales)
    return total,average

sales = [2000,4000,3000,1000,6000]

total_sales, average_sales = sales_analysis(sales)

print("Total Sales:", total_sales)
print("Average Sales:", average_sales)
```

```python
#Create a small Python program that reads sales data and calculates basic totals.
```

```python
sales_data = [
    {"product": "Laptop", "quantity": 2, "price": 50000},
    {"product": "Mouse", "quantity": 5, "price": 500},
    {"product": "Keyboard", "quantity": 3, "price": 1500},
    {"product": "Monitor", "quantity": 2, "price": 12000}
]

total_sales = 0
total_quantity = 0

for item in sales_data:

    revenue = item["quantity"] * item["price"]

    total_sales += revenue
    total_quantity += item["quantity"]

print("Total Quantity:", total_quantity)
print("Total Sales:", total_sales)
```

```python
highest_revenue = 0
highest_product = ""

for item in sales_data:
    revenue = item["quantity"] * item["price"]

    if revenue > highest_revenue:
        highest_revenue = revenue
        highest_product = item["product"]

print("Highest Revenue Product:", highest_product)
print("Highest Revenue:", highest_revenue)
```
