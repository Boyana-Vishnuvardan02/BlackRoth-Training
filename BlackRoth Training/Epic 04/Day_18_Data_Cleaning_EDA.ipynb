```python
Q1 = df["Revenue"].quantile(0.25)
Q3 = df["Revenue"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["Revenue"] < lower_bound) |
    (df["Revenue"] > upper_bound)
]

print("Number of outliers:", len(outliers))
```

```python
## EDA Questions
```

```python
#Which product generates the most revenue? 
```

```python
Product_revenue = df.groupby("Product")["Revenue"].sum()
print("Product_Name:",Product_revenue.idxmax(),",","Revenue:",Product_revenue.max())
```

```python
#Which region performs best? 
```

```python
Best_Region = df.groupby("State")["Revenue"].sum()
print("Best_Region:",Best_Region.idxmax(),",","Revenue:",Best_Region.max())
```

```python
#Which month has the highest sales? 
```

```python
Monthly_Sales = df.groupby("Month")["Revenue"].sum()

print("Top_Month:",Monthly_Sales.idxmax(),",","Top_Month:",Monthly_Sales.max())
```

```python
#What is the average order value?
```

```python
Total_Revenue = df["Revenue"].sum()
Total_Orders = df["Order_ID"].nunique()

Average_Order_Value = Total_Revenue / Total_Orders

print("Average_Order_Value:", Average_Order_Value)
```

```python
#Which category has the highest quantity?
```

```python
Highest_Category_By_Quantity = df.groupby("Category")["Quantity"].sum()
print("Highest_Category:",Highest_Category_By_Quantity.idxmax(),",","Quantity:",Highest_Category_By_Quantity.max())
```

```python
#Are there unusual values or outliers?
```

```python
Q1 = df["Revenue"].quantile(0.25)
Q3 = df["Revenue"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[
    (df["Revenue"] < lower) |
    (df["Revenue"] > upper)
]

print("Number of revenue outliers:", len(outliers))
```
