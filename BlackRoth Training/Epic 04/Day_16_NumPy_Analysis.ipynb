```python
#Create sales data and calculate total sales
```

```python
sales = np.array([10000,15000,20000,25000,30000])

print(np.sum(sales))
```

```python
#Find sales above ₹20,000
```

```python
print(sales[sales>20000])
```

```python
#Increase all sales by 10%.
```

```python
new_sales = sales * 1.10

print(new_sales)
```

```python
#Classify sales as High or Low where sales ≥ ₹20,000 is High
```

```python
print(np.where(new_sales>=20000,"high","low"))
```

```python
#Mini Data Analyst Project
```

```python
sales = np.array([12000,15000,18000,22000,25000,19000,30000,28000,17000,35000])

print("Sales Data:", sales)

print("Total Sales:", np.sum(sales))
print("Average Sales:", np.mean(sales))
print("Minimum Sales:", np.min(sales))
print("Maximum Sales:", np.max(sales))
print("Median Sales:", np.median(sales))
print("Standard Deviation:", np.std(sales))

high_sales = sales[sales > 20000]

print("High Value Sales:", high_sales)
print("Number of High Value Sales:", len(high_sales))
print("Average High Value Sales:", np.mean(high_sales))

increased_sales = sales * 1.10

print("Sales After 10% Increase:", increased_sales)
```
